# Synthetic Fraud Data Exploration

**Synthetic demonstration data. Not valid for real payment decisions.**

```python
from ml.data.generate import create_synthetic
import matplotlib.pyplot as plt
df = create_synthetic(4000)
print(df.describe(include='all').to_string())
print('Synthetic fraud rate:', df.is_fraud.mean())
df.groupby('merchant_category').is_fraud.mean().plot.bar(title='Synthetic fraud label rate by category'); plt.show()
```
