# Training-only Preprocessing

**Synthetic demonstration data. Not valid for real payment decisions.**

```python
from ml.data.generate import create_synthetic
from ml.data.feature_engineering import make_preprocessor
from ml.config import FEATURES
df = create_synthetic()
split = int(.6 * len(df))
transformer = make_preprocessor()
X_train = transformer.fit_transform(df[FEATURES].iloc[:split])
X_validation = transformer.transform(df[FEATURES].iloc[split:int(.8*len(df))])
print(X_train.shape, X_validation.shape)
```
