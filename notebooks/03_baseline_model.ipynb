# Baseline Fraud Classifier

**Synthetic demonstration data. Not valid for real payment decisions.**

```python
from ml.data.generate import create_synthetic
from ml.config import FEATURES
from ml.data.feature_engineering import make_preprocessor
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score
df = create_synthetic()
a=int(len(df)*.6); b=int(len(df)*.8)
model = make_pipeline(make_preprocessor(), LogisticRegression(class_weight='balanced',max_iter=600))
model.fit(df[FEATURES].iloc[:a],df.is_fraud.iloc[:a])
p = model.predict_proba(df[FEATURES].iloc[a:b])[:,1]
print('Validation PR-AUC:', average_precision_score(df.is_fraud.iloc[a:b],p))
```
